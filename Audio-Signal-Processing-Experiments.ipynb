```python
def amplitude_envelope(signal,frame_size,hop_length):
  frames_arr=[]
  for i in range(0,len(signal),hop_length):
    data_frame=signal[i:i+frame_size]
    frames_arr.append(max(data_frame))
  return np.array(frames_arr)

```

```python
FRAME_SIZE = 1024
HOP_LENGTH = 512
res=amplitude_envelope(debussy,FRAME_SIZE,HOP_LENGTH)
res2=amplitude_envelope(redhot,FRAME_SIZE,HOP_LENGTH)
res3=amplitude_envelope(duke,FRAME_SIZE,HOP_LENGTH)
```

```python
def RMS(signal,frame_size,hop_length):
  frames_arr=[]
  final_arr=[]
  for i in range(0,len(signal),hop_length):
    data_frame=signal[i:i+frame_size]
    rms_val=np.sqrt(np.mean(np.power(data_frame,2)))
    final_arr.append(rms_val)
  return np.array(final_arr)
```

```python
FRAME_SIZE = 1024
HOP_LENGTH = 512

rms=librosa.feature.rms(y=debussy,frame_length=FRAME_SIZE,hop_length=HOP_LENGTH)[0]
rms2=librosa.feature.rms(y=redhot,frame_length=FRAME_SIZE,hop_length=HOP_LENGTH)[0]
rms3=librosa.feature.rms(y=duke,frame_length=FRAME_SIZE,hop_length=HOP_LENGTH)[0]
```

```python
FRAME_SIZE = 1024
HOP_LENGTH = 512

rms4=RMS(debussy,FRAME_SIZE,HOP_LENGTH)
rms5=RMS(redhot,FRAME_SIZE,HOP_LENGTH)
rms6=RMS(duke,FRAME_SIZE,HOP_LENGTH)
```

```python
def sampels_to_frames(sampels,frame_size,hop_size):
  frames=[]
  for i in range(0,len(sampels),hop_size):
    frame_sampels=sampels[i:i+frame_size]
    paded_frame=np.pad(frame_sampels,(0,frame_size-len(frame_sampels)),mode="constant")
    windowd_frame=np.hanning(len(paded_frame))*paded_frame
    frames.append(windowd_frame)
  return np.array(frames)

FRAME_SIZE=2048
HOP_SIZE=512
scale_frames=sampels_to_frames(scale,FRAME_SIZE,HOP_SIZE)
# np.fft.fft(scale_frames[1])[:1025][0]
stft_scale=librosa.stft(y=scale,n_fft=FRAME_SIZE,hop_length=HOP_SIZE,center=False)
# x=stft_scale.T
```

```python
stft_scale=librosa.stft(y=scale,n_fft=FRAME_SIZE,hop_length=HOP_SIZE,center=False)
Y_scale = np.abs(stft_scale) ** 2
y_scale_db=librosa.power_to_db(Y_scale)
```

```python
def plot_spectrogram(Y, sr=22050, hop_length=512, y_axis="linear"):
    plt.figure(figsize=(25, 10))
    librosa.display.specshow(Y,
                             sr=sr,
                             hop_length=hop_length,
                             x_axis="time",
                             y_axis=y_axis)
    plt.colorbar(format="%+2.f")

plot_spectrogram(y_scale_db,y_axis="log")
```
