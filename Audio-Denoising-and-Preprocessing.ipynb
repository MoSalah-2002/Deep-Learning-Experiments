```python
import soundfile as sf
import numpy as np
from nara_wpe.wpe import wpe_v8
from scipy.signal import stft, istft

# Load audio
audio, sr = sf.read("/content/audiomass-output.mp3")

# Convert stereo to mono if needed
if audio.ndim > 1:
    audio = audio.mean(axis=1)

# STFT
_, _, Zxx = stft(
    audio,
    fs=sr,
    nperseg=512,
    noverlap=256
)

# WPE expects shape (channels, frequency_bins, frames)
Y = Zxx[np.newaxis, :, :]

# Apply WPE
X = wpe_v8(
    Y,
    taps=10,
    delay=3,
    iterations=5
)

# Back to time domain
_, enhanced = istft(
    X[0],
    fs=sr,
    nperseg=512,
    noverlap=256
)

sf.write("dereverb.wav", enhanced.astype(np.float32), sr)
```
