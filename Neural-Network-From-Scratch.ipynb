```python
import numpy as np
```

```python
def actvation_function(x,thresh):
  o=0
  if x>=thresh:
    o=1
  else:
    o=0
  return o
```

```python
def forward_prob(input_tuple_list,true_output,w1,w2,w_bais):
  net=0
  c=0
  error_list=[]
  for tuble in input_tuple_list:
    x1=tuble[0]
    x2=tuble[1]
    b=tuble[2]
    net=x1*w1+x2*w2+b*w_bais
    net=actvation_function(net,0.1)
    error=true_output[c]-net
    error_list.append(error)
    c=c+1
    # print(net,error)
    if error!=0:
      w1=w1+(error*x1)
      w2=w2+(error*x2)
      w_bais=w_bais+(error*b)

  return (w1,w2,w_bais,error_list)
```

```python
def predict(Weights,input):
   x1=input[0]
   x2=input[1]
   b=input[2]
   w1=Weights[0][0]
   w2=Weights[0][1]
   w_bais=Weights[0][2]
   net=x1*w1+x2*w2+b*w_bais
   net=actvation_function(net,0.1)
   return net
```

```python
x1_list=[2,-3,3,-1,7,0,0]
x2_list=[3,3,4,2,2,1,-2]
t=[0,1,0,1,0,1,1]

# x1_list=[0,0,1,1]
# x2_list=[0,1,0,1]
# t=[0,1,1,1]

tuble=[]
for x1,x2 in zip(x1_list,x2_list):
  tuble.append((x1,x2,1))


final_Weights=[]
Weights=forward_prob(tuble,t,0.1,0.2,-0.2)
for i in range(100):
  if  1 in Weights[3] or -1 in Weights[3]:
    Weights=forward_prob(tuble,t,Weights[0],Weights[1],Weights[2])
  else:
    final_Weights.append(Weights)
    final_Weights.append(f"we stop at itration {i+1}")
    break
# print(final_Weights)

print(predict(final_Weights,(-2,10,1)))
```
