```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import tensorflow as tf
import keras
from keras.applications.vgg16 import VGG16
from keras.applications.vgg16 import preprocess_input
from keras.preprocessing import image
```

```python
model=VGG16(weights='imagenet')
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen=ImageDataGenerator(preprocessing_function=keras.applications.vgg16.preprocess_input).flow_from_directory(directory='/content/asl_alphabet_train/asl_alphabet_train',target_size=(224,224),batch_size=32)
# test_datagen=ImageDataGenerator(preprocessing_function=keras.applications.vgg16.preprocess_input).flow_from_directory(directory='/content/asl_alphabet_test',target_size=(224,224),batch_size=32)
```

```python
img,lable=next(train_datagen)
labels=train_datagen.class_indices
rev_labels={v:k for k,v in labels.items()}
# rev_labels[np.argmax(lable[0])]
```

```python
fined_model=keras.models.Sequential()
for layer in model.layers[:-1]:
  fined_model.add(layer)
for layer in fined_model.layers:
  layer.trainable=False

fined_model.add(keras.layers.Dense(len(labels),activation='softmax'))
```

```python
# fined_model.summary()

fined_model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
fined_model.fit(train_datagen,epochs=10)
```

```python
def pre_process(img):
  img=image.load_img(img,target_size=(224,224))
  img=image.img_to_array(img)
  img=np.expand_dims(img,axis=0)
  img=preprocess_input(img)
  return img
```

```python
import os

images=[]
images_name=[]
for i in os.listdir('/content/asl_alphabet_test/asl_alphabet_test'):
  images_name.append(i)
  im_path=os.path.join('/content/asl_alphabet_test/asl_alphabet_test',i)
  img=pre_process(im_path)
  images.append(img)
```

```python
predctions=fined_model.predict(images[15])
```
